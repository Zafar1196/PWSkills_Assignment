````python
# File operation
import os

filename = "test.txt"

with open(filename, "w", encoding="utf-8") as file:
    file.write("Hello, World!")
    file.write("This is a line of first coding")

print(f"Created and wrote initial data to '{filename}'.")


# --- 2. Appending Data to a File ---
# 'a' mode adds data to the end without deleting existing content
with open(filename, "a", encoding="utf-8") as file:
    file.write("Appended line 1")
    file.write("Appended line 2")

print("Appended new lines.")


# --- 3. Reading from a File ---
# Read line by line (memory-efficient for large files)
print("Reading line by line ---")
with open(filename, "r", encoding="utf-8") as file:
    for line_number, line in enumerate(file, 1):
        print(f"Line {line_number}: {line.strip()}")

# Read all contents at once into a single string
with open(filename, "r", encoding="utf-8") as file:
    content = file.read()

if os.path.exists(filename):
    os.remove(filename)
    print(f"Cleaned up: Removed '{filename}'.")

##### Data processing

import csv

sales_data = [
    ["Name", "Department", "Sales"],
    ["Zafar", "Engineering", 7500],
    ["Quasim", "Marketing", 4200],
    ["Saamiya", "Engineering", 8100],
    ["Seraj", "Marketing", 5600],
    ["Vanchung", "Sales", 6300]
]

csv_filename = "sales.csv"

# Write data to CSV
with open(csv_filename, "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(sales_data)

# Read and Process CSV Data
total_sales = 0
department_sales = {}

with open(csv_filename, "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)

    for row in reader:
        name = row["Name"]
        dept = row["Department"]
        amount = int(row["Sales"])

        # Calculate totals
        total_sales += amount
        department_sales[dept] = department_sales.get(dept, 0) + amount

print(f"Total Sales: ${total_sales:,}")
print("\nSales by Department:")
for dept, total in department_sales.items():
    print(f"  - {dept}: ${total:,}")
````
