```python
## Create folder
Ansible-Project
  > inventory.ini
  > main.yml
  > index.html


  ##landing Page

  <!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <title>Welcome</title>
    <style>
        body { font-family: sans-serif; text-align: center; margin-top: 100px; background: #f4f4f9; }
        h1 { color: #333; }
    </style>
</head>
<body>
    <h1>Managed by Ansible</h1>
    <p>This web server ({{ ansible_hostname }}) was automatically configured using Ansible.</p>
</body>
</html>

### write play book for nginx configuration

- name: Deploy and Configure Nginx Web Server
  hosts: webservers
    web_server_port: 80

  tasks:
    - name: Ensure apt cache is up to date (Debian/Ubuntu)
      ansible.builtin.apt:

    - name: Install Nginx
      ansible.builtin.package:
        name: nginx
        state: present

    - name: Ensure UFW allows HTTP traffic (Debian/Ubuntu)
      community.general.ufw:
        rule: allow
        port: "{{ web_server_port }}"
        proto: tcp
      when: ansible_os_family == "Debian"

    - name: Deploy custom web page
      ansible.builtin.template:
        src: index.html
        dest: /var/www/html/index.html
        owner: www-data
        group: www-data
        mode: '0644'
      notify: Reload Nginx

    - name: Ensure Nginx is running and enabled on boot
      ansible.builtin.service:
        name: nginx
        state: started
        enabled: true

  handlers:
    - name: Reload Nginx
      ansible.builtin.service:
        name: nginx
        state: reloaded

## Verify the connection with host
             ansible webserver  -i inventory.ini
## Run the playbook

             ansible-playbook - inventory.ini main.yml
```
