Use Docker Compose to define and run multi-container applications?

 * Once the YML file is ready and set up use these commands from directory where docker file xyz.yml file is located.

 ##### Start the Application
        * docker compose up


 ##### Rebuild the image

   *      docker compose up --build   


 ##### Stop and Remove the container
   *         docker compose down


 ##### View running container ans status
   *      docker compose ps

##### View logs file

 *      docker compose logs -f
