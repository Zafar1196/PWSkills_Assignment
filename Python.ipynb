3. Compare and contrast mutable and immutable objects in Python with examples
   - Mutable → Can be changed after creation.
   - Immutable → Cannot be changed after creation.

 * Examples of Mutable objects: list, dictionary, set > these all are can be change after its created
 * Examples of immutable objects: int, float, string, tuple > these all are cannot be changed after created


```python
#Mutable example (List)
numbers = [10, 20, 30]
numbers.append(40)

print(numbers)  #output will [10,20,30,40]

#Immutable example (Strings)

name = "zafar"

name = name + " Zafar Quasim"

print(name)  #output will be [Zafar quasim]
```

5. Explain the concept of type casting in Python with examples.

   - Type casting is the process of converting one data type into another in Python using built-in functions like int(), Float(), str(), Boolean() etc.
      
    - why use type casting:
       - To convert one data type to another data type
       - To perform calculation on different data types
       - To accept user input and process it correctly
       - To avoid type related errors.
     - Example:
       - integer to float >
          - a = 10 , print(float(a)) then output will be 10.0
       - Float to integer >
        -  b = 5.8 , print(int(b)) then output will be 5

       - String to integer >
        - c = "20" , print(int(c)) then output will be 20
