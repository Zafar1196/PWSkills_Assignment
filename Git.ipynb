#### 5. Tasks to Be Performed:
1. Create a Git Flow workflow architecture on Git
2. Create all the required branches
3. starting from e feature branch, push the branch to the master, following the architecture
4. Push an urgent.txt on master using hotfix

```python
## 1. Initialize the repo and create branch
mkdir git-flow-demo
cd git-flow-demo
git init

# Create initial commit on master
touch README.md
git add README.md
git commit -m "Initial commit on master"

# Create and switch to develop branch
git checkout -b develop


## 2. Create the Required Supporting Branches

git checkout -b feature/new-feature develop

# Create a release branch off develop
git checkout -b release/v1.0.0 develop


## 3. Work on Feature and Promote it to Master

 # --- Step A: Work on Feature ---
git checkout feature/new-feature
touch feature.txt
git add feature.txt
git commit -m "Add feature code"

# --- Step B: Merge Feature into Develop ---
git checkout develop
git merge --no-ff feature/new-feature
git branch -d feature/new-feature # Clean up feature branch

# --- Step C: Prepare Release from Develop ---
git checkout -b release/v1.0.0 develop


# --- Step D: Merge Release into Master ---
git checkout master
git merge --no-ff release/v1.0.0
git tag -a v1.0.0 -m "Release v1.0.0"

# --- Step E: Sync Release Back to Develop & Cleanup ---
git checkout develop
git merge --no-ff release/v1.0.0
git branch -d release/v1.0.0

## 4. Push urgent.txt to Master Using a Hotfix Branch

git checkout master
git checkout -b hotfix/urgent-fix

# --- Step B: Add urgent.txt ---
touch urgent.txt
git add urgent.txt
git commit -m "Add urgent.txt hotfix"

# --- Step C: Merge Hotfix into Master ---
git checkout master
git merge --no-ff hotfix/urgent-fix
git tag -a v1.0.1 -m "Hotfix release v1.0.1"

# --- Step D: Merge Hotfix into Develop ---
git checkout develop
git merge --no-ff hotfix/urgent-fix

# --- Step E: Delete Hotfix Branch ---
git branch -d hotfix/urgent-fix
```
