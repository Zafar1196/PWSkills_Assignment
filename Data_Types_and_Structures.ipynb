7. What is a hash table, and how does it relate to dictionaries in Python?
   - A hash table is a fundamental data structure designed to store data in key-value pairs. It is designed to perform core operations—like inserting, deleting, and searching for data—almost instantly, achieving a time.
  ##### How hashing work
   - Imagine you run a massive warehouse with thousands of boxes. If you stack them randomly, finding a specific box requires checking them one by one. A hash table solves this by using a mathematical formula to determine exactly which shelf a box belongs on before it even enters the warehouse.A standard hash table consists of two main parts The Array A fixed-size internal array where the data values are stored.The Hash Function: A deterministic algorithm that takes a key a string and converts it into an integer .This integer is then mapped to a specific index in the array using the modulo operator.
  
  ##### How relate with Dictionaries
  - Every time you interact with a dictionary in Python, you are directly utilizing an underlying hash table. However, Python implements this concept with a highly specialized layout to solve two classic hash table limitation hash collisions and memory overhead.
  - Handling Hash Collisions
Because there are infinite possible keys but a finite number of array slots, two different keys will occasionally map to the exact same index. This is called a hash collision

15. Can you modify the elements of a tuple? Explain why or why not
 - We can not modify the elements of tuple. If you attempt to change, add, or delete an item in a tuple after it has been created, Python will block the operation and give errors like TypeError.
 - Python tuples are designed to be immutable. When Python allocates memory for a tuple, it allocates a fixed-size block that contains permanent references to specific objects. The length of the tuple is locked, and the pointers inside those slots cannot be overwritten or swapped out.
 - Exeption is If you place a mutable object like a list or a dictionary inside a tuple, you can still modify the contents of that nested object. The tuple hasn't changed—it is still pointing to the exact same list in memory—but the list inside it has been mutated.

19. Why are dictionaries considered unordered, and how does that affect data retrieval
    - Dictionaries based on a hash table When you add a key-value pair, Python runs the key through a hash function to calculate a random-looking memory index.
  Because the insertion index depends entirely on the mathematical output of the hash function, the items were scattered across internal memory buckets with no regard for the order in which you typed them.

- Because data retrieval bypasses positioning, you don't need to know where an item is to find it. In a list of 1 million items, retrieving the last item requires jumping through a sequence of indices or searching linearly. In a dictionary, retrieving the key bypasses the order entirely. Python hashes the key and grabs the value instantly, making retrieval speed entirely independent of how many items are in the dictionary or where the item sits sequentially.
