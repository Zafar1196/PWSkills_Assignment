2. What is the difference between Hard Link & Soft Link
  * Hard Link and a Soft Link (also known as a Symbolic Link or Symlink) in Linux, it helps to understand how the file system tracks files
  Every file in Linux consists of two parts:
  The Inode: An index number containing the file's metadata (permissions, owner, size) and pointing to the actual data stored on the disk.
  The File Name: A human-readable pointer/label linked to that specific inode.

#### Hard Link :-
  * A hard link is an additional file name pointing to the exact same inode as the original file. The file system treats both the original name and the hard link as equal, direct pathways to the same underlying data blocks.
  * If you delete the original file, the data is NOT lost. The hard link continues to point to the data. The data is only removed from the disk when all hard links pointing to that inode are deleted.

#### Soft Link :-
  * A soft link is an entirely new file with its own unique inode that contains a path pointing to another file's name.A soft link is an entirely new file with its own unique inode that contains a path pointing to another file's name. If you delete the original file, the soft link remains, but it becomes broken (pointing to a target path that no longer exists).


3. What is a Kernel in Linux?
  * The kernel is the primary core of the Linux operating system. It acts as an invisible mediator sitting directly between your computer's hardware and the software applications running on it. Without a kernel, your applications (like web browsers, text editors, or media players) would have no way of talking to your physical hardware (like your CPU, RAM, or hard drive)

 #### Task Of Kernel :-

  * Memory Management: It keeps track of how much memory (RAM) is available, allocates memory to specific programs when they open, and reclaims that memory when they close.

  * Process Management (CPU Scheduling): It decides which running program gets to use the CPU, for how long, and in what order—allowing your computer to multitask seamlessly without crashing.

  * Device Drivers: It acts as a universal interpreter between your hardware devices (mouse, keyboard, Wi-Fi card, graphics card) and the applications trying to use them.

 * System Calls & Security: It controls hardware access     permissions, ensuring that an app cannot improperly access memory used by another application or tamper with protected system files.
