```python
## Steps for Installing Ansible

# Update package
sudo apt update
sudo apt install -y

# Add Ansible's
sudo apt-add-repository --yes --update ppa:ansible/ansible

# Install Ansible
sudo apt install -y ansible


### To verify The installation
ansible --V


### Create Work directory

mkdir ~/ansible-demo
cd ~/ansible-demo


## Define the host node


[local]
localhost ansible_connection=local

## Create file name
ansible_demo.yml

---
- name: Configure Web Server on Localhost
  hosts: local

  tasks:
    - name: install nginx
      apt:
        name: nginx
        state: present
        update_cache: yes

    - name: Create index.html page
      copy:
        content: "<h1>Hello ansible please create my first playbook</h1>
        dest: /var/www/html/index.html

    - name: Nginx service is running and enabled on boot
      systemd:
        name: nginx
        state: started
        enabled: yes

 ### Execute the ansible playbook

        ansible-playbook -i host ansible_demo.yml


 ### verify the Deployment

   http://localhost # on the local machine
```
